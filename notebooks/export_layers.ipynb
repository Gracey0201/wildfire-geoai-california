```python
from pathlib import Path
import shutil
```

```python
# =========================================================
# PATHS
# =========================================================

PROJECT_ROOT = Path.cwd().parent

ALIGNED_DIR = (
    PROJECT_ROOT /
    "data" /
    "processed" /
    "aligned"
)

FEATURE_DIR = (
    PROJECT_ROOT /
    "data" /
    "features"
)

VECTOR_DIR = (
    PROJECT_ROOT /
    "data" /
    "processed" /
    "reprojected"
)

OUTPUT_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "figures" /
    "layers"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print(OUTPUT_DIR)
```

```python
# =========================================================
# CLEAN OUTPUT FOLDER
# =========================================================

for file in OUTPUT_DIR.glob("*"):

    if file.is_file():

        file.unlink()

print("Output folder cleaned")
```

```python
# =========================================================
# EXPORT ALL LAYERS
# =========================================================

layers = {

    # Base layers
    ALIGNED_DIR / "dem_aligned.tif":
        OUTPUT_DIR / "dem.tif",

    ALIGNED_DIR / "landcover_aligned.tif":
        OUTPUT_DIR / "landcover.tif",

    ALIGNED_DIR / "precipitation_aligned.tif":
        OUTPUT_DIR / "precipitation.tif",

    # Feature layers
    FEATURE_DIR / "slope.tif":
        OUTPUT_DIR / "slope.tif",

    FEATURE_DIR / "twi.tif":
        OUTPUT_DIR / "twi.tif",

    FEATURE_DIR / "distance_to_roads.tif":
        OUTPUT_DIR / "distance_to_roads.tif",

    FEATURE_DIR / "distance_to_settlements.tif":
        OUTPUT_DIR / "distance_to_settlements.tif",

    # Vector layers
    VECTOR_DIR / "roads_3310.geojson":
        OUTPUT_DIR / "roads.geojson",

    VECTOR_DIR / "wildfire_3310.geojson":
        OUTPUT_DIR / "wildfire_history.geojson"

}

for source, destination in layers.items():

    shutil.copy2(
        source,
        destination
    )

    print(
        f"Exported: {destination.name}"
    )
```

```python
# =========================================================
# VERIFY EXPORTS
# =========================================================

print("\n===================================")
print("EXPORTED LAYERS")
print("===================================\n")

for file in sorted(
    OUTPUT_DIR.iterdir()
):

    print(file.name)

print(
    f"\nTotal files: "
    f"{len(list(OUTPUT_DIR.iterdir()))}"
)
```
