```python
import matplotlib.pyplot as plt
import rioxarray as rxr
from matplotlib.colors import ListedColormap
import numpy as np
import pandas as pd
```

```python
labels = rxr.open_rasterio(
    "../data/training/labels.tif"
).squeeze()

unique, counts = np.unique(
    labels.values,
    return_counts=True
)

print("\nLabel Summary\n")

for value, count in zip(
    unique,
    counts
):

    if value == 0:

        print(
            f"Unburned: {count:,}"
        )

    elif value == 1:

        print(
            f"Burned: {count:,}"
        )

cmap = ListedColormap(
    [
        "lightgray",
        "red"
    ]
)

fig, ax = plt.subplots(
    figsize=(8, 6)
)

im = ax.imshow(
    labels,
    cmap=cmap
)

cbar = plt.colorbar(
    im,
    ax=ax,
    ticks=[0, 1]
)

cbar.ax.set_yticklabels(
    [
        "Unburned",
        "Burned"
    ]
)

ax.set_title(
    "Wildfire Labels"
)

plt.show()
```
