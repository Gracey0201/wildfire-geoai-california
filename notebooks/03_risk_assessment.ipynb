# Wildfire Exposure and Risk Assessment

This notebook examines wildfire exposure and risk across Butte County, California.

Objectives:

Visualize wildfire exposure patterns
Visualize wildfire risk patterns
Summarize exposure and risk statistics
Examine the distribution of exposure and risk values
Identify areas where wildfire impacts may be most severe
Support mitigation planning and decision-making

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import rasterio

from pathlib import Path
```

```python
PROJECT_ROOT = Path("..")

MAPS_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "maps"
)

FIGURES_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "figures"
)

EXPOSURE_TIF_PATH = (
    MAPS_DIR /
    "wildfire_exposure.tif"
)

RISK_TIF_PATH = (
    MAPS_DIR /
    "wildfire_risk.tif"
)
```

```python
## Load raster

with rasterio.open(
    EXPOSURE_TIF_PATH
) as src:

    exposure = src.read(1)

    nodata = src.nodata

exposure = np.where(
    exposure == nodata,
    np.nan,
    exposure
)

valid_exposure = exposure[
    np.isfinite(exposure)
]
```

## Wildfire Exposure Map

```python
vmax = np.nanpercentile(
    valid_exposure,
    99
)

plt.figure(figsize=(10,8))

img = plt.imshow(
    exposure,
    cmap="viridis",
    vmin=0,
    vmax=vmax
)

cbar = plt.colorbar(
    img
)

cbar.set_label(
    "Wildfire Exposure Index"
)

plt.title(
    "Wildfire Exposure Surface"
)

plt.axis("off")

output_path = (
    FIGURES_DIR /
    "wildfire_exposure_map.png"
)

plt.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight"
)

print(
    f"Saved: {output_path}"
)

plt.show()
```

### Wildfire Exposure Map Interpretation

The wildfire exposure surface identifies locations where wildfire hazards intersect with human populations, infrastructure, and developed landscapes throughout Butte County.

Exposure is not evenly distributed across the county. Instead, elevated exposure values are concentrated within developed areas and wildland–urban interface (WUI) zones where people and assets are most likely to be affected by wildfire events. Large portions of the county exhibit relatively low exposure, reflecting areas with limited population density and fewer built assets.

The spatial pattern indicates that potential wildfire impacts are concentrated within specific communities and development corridors rather than being uniformly distributed across the landscape.

## Exposure Statistics

```python
exposure_stats = pd.DataFrame(
    {
        "Statistic": [
            "Minimum",
            "Maximum",
            "Mean",
            "Median",
            "Standard Deviation"
        ],
        "Value": [
            valid_exposure.min(),
            valid_exposure.max(),
            valid_exposure.mean(),
            np.median(valid_exposure),
            valid_exposure.std()
        ]
    }
)

exposure_stats["Value"] = exposure_stats["Value"].round(3)

exposure_stats
```

### Exposure Statistics Interpretation

The wildfire exposure index ranges from 0.00 to 0.95, indicating substantial variation in exposure levels across Butte County.

The mean exposure value of 0.119 and median value of 0.106 suggest that most locations within the county experience relatively low exposure to wildfire impacts. The close agreement between the mean and median indicates that low-exposure areas dominate the landscape.

The maximum exposure value of 0.953 identifies localized areas where populations, infrastructure, and assets are highly concentrated and therefore potentially more vulnerable to wildfire impacts.

The standard deviation of 0.116 indicates moderate spatial variability in exposure across the county. This variability reflects the uneven distribution of communities and infrastructure, with exposure concentrated in developed areas and Wildland–Urban Interface (WUI) zones while large portions of the county remain sparsely populated and exhibit low exposure levels.

Overall, the statistics demonstrate that wildfire exposure is not uniformly distributed across Butte County but is concentrated within specific areas where wildfire events could have greater consequences for people and assets.

## Exposure Histogram

```python
plt.figure(figsize=(8,5))

plt.hist(
    valid_exposure,
    bins=30
)

plt.xlabel("Exposure Index")
plt.ylabel("Pixel Count")
plt.title("Distribution of Wildfire Exposure")

plt.tight_layout()
plt.show()
```

### Exposure Distribution Interpretation

The histogram reveals a strongly right-skewed distribution of wildfire exposure values across Butte County.

Most pixels are concentrated within the lower exposure range (approximately 0.00–0.20), with the highest concentration occurring near an exposure index of 0.10. This pattern indicates that the majority of the county contains relatively low levels of exposed population and infrastructure.

A smaller proportion of pixels exhibit moderate to high exposure values, producing a long upper tail in the distribution. These high-exposure areas correspond to developed communities, transportation corridors, and Wildland–Urban Interface (WUI) zones where wildfire impacts could affect larger concentrations of people and assets.

Overall, the distribution demonstrates that wildfire exposure is spatially concentrated rather than uniformly distributed across the county.

## Load and clean raster for Risk Assessment

```python
with rasterio.open(
    RISK_TIF_PATH
) as src:

    risk = src.read(1)

    nodata = src.nodata

risk = np.where(
    risk == nodata,
    np.nan,
    risk
)

valid_risk = risk[
    np.isfinite(risk)
]
```

## Wilfire Risk Map

```python
with rasterio.open(
    RISK_TIF_PATH
) as src:

    risk = src.read(1)

    nodata = src.nodata

risk = np.where(
    risk == nodata,
    np.nan,
    risk
)

vmax = np.nanpercentile(
    risk,
    99
)

plt.figure(figsize=(10,8))

img = plt.imshow(
    risk,
    cmap="YlOrRd",
    vmin=0,
    vmax=vmax
)

cbar = plt.colorbar(
    img
)

cbar.set_label(
    "Wildfire Risk Index"
)

plt.title(
    "Wildfire Risk Surface"
)

plt.axis("off")

output_path = (
    FIGURES_DIR /
    "wildfire_risk_map.png"
)

plt.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight"
)

print(
    f"Saved: {output_path}"
)

plt.show()
```

### Wildfire Risk Map Interpretation

The wildfire risk map identifies areas where wildfire susceptibility overlaps with exposed populations and infrastructure to produce elevated overall risk.

High-risk areas are concentrated primarily within the central and south-central portions of Butte County, where wildfire-prone environments intersect with developed areas and Wildland–Urban Interface (WUI) zones. These locations exhibit the highest potential for adverse wildfire impacts due to the combined influence of hazard and exposure.

In contrast, large portions of the county display low risk values despite varying levels of wildfire susceptibility. This occurs because many wildfire-prone locations contain relatively few people, structures, or assets that could be affected by wildfire events.

The spatial pattern demonstrates that wildfire risk is not solely determined by wildfire likelihood but also by the presence of exposed communities and infrastructure. Consequently, risk management efforts should prioritize areas where elevated hazard coincides with high levels of exposure.

## Risk Statistics Table

```python
risk_stats = pd.DataFrame(
    {
        "Statistic": [
            "Minimum",
            "Maximum",
            "Mean",
            "Median",
            "Standard Deviation"
        ],
        "Value": [
            valid_risk.min(),
            valid_risk.max(),
            valid_risk.mean(),
            np.median(valid_risk),
            valid_risk.std()
        ]
    }
)

risk_stats["Value"] = (
    risk_stats["Value"]
    .round(3)
)

risk_stats
```

### Risk Statistics Interpretation

The wildfire risk index ranges from 0.000 to 0.560, indicating substantial spatial variation in wildfire risk across Butte County.

The mean risk value of 0.008 and median value of 0.000 indicate that most locations within the county experience very low wildfire risk. The median of zero suggests that more than half of the mapped area has negligible risk under the current modeling framework.

The maximum risk value of 0.560 identifies localized hotspots where wildfire susceptibility, exposure, and vulnerability intersect to produce elevated overall risk.

The standard deviation of 0.018 indicates that risk values are concentrated near the lower end of the distribution, with relatively few areas exhibiting moderate to high risk.

Overall, the statistics demonstrate that wildfire risk is highly concentrated within specific portions of the county rather than being uniformly distributed across the landscape. These high-risk areas represent locations where wildfire occurrence could result in the greatest impacts on people, infrastructure, and community assets.

## Risk Histogram

```python
plt.figure(figsize=(8,5))

plt.hist(
    valid_risk,
    bins=30
)

plt.xlabel(
    "Risk Index"
)

plt.ylabel(
    "Pixel Count"
)

plt.title(
    "Distribution of Wildfire Risk"
)

plt.tight_layout()

output_path = (
    FIGURES_DIR /
    "risk_histogram.png"
)

plt.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight"
)

print(
    f"Saved: {output_path}"
)

plt.show()
```

### Risk Histogram Interpretation

The histogram reveals a strongly right-skewed distribution of wildfire risk values.

Most pixels are concentrated near zero, indicating that the majority of Butte County experiences relatively low overall wildfire risk. As risk values increase, the number of pixels declines rapidly, producing a long tail toward higher risk values.

This pattern suggests that elevated wildfire risk is confined to a relatively small proportion of the landscape. These localized high-risk areas correspond to locations where hazardous wildfire conditions overlap with exposed populations, infrastructure, and other assets.

The distribution highlights the importance of targeted risk reduction strategies, as a limited number of locations account for a disproportionately large share of overall wildfire risk.
