# Wildfire Susceptibility Mapping

This notebook presents the spatial outputs generated by the Random Forest wildfire susceptibility model for Butte County, California.

Objectives:

- Visualize wildfire susceptibility patterns
- Classify susceptibility into risk categories
- Identify high-risk wildfire zones
- Summarize susceptibility distribution across the county

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image
from IPython.display import display

import rasterio

from pathlib import Path
```

```python
PROJECT_ROOT = Path("..")

MAPS_DIR = (
    PROJECT_ROOT /
    "outputs" /
    "maps"
)

SUSCEPTIBILITY_PATH = (
    MAPS_DIR /
    "wildfire_susceptibility.tif"
)

SUSCEPTIBILITY_CLASSES_PATH = (
    MAPS_DIR /
    "wildfire_susceptibility_classes.tif"
)

SUSCEPTIBILITY_PNG_PATH = (
    MAPS_DIR /
    "wildfire_susceptibility.png"
)
```

```python
display(
    Image.open(
        SUSCEPTIBILITY_PNG_PATH
    )
)
```

### Susceptibility Map Interpretation

The susceptibility map illustrates the spatial probability of wildfire occurrence across Butte County.

Areas classified with higher susceptibility values represent locations where environmental and human-related conditions are more favorable for wildfire occurrence. Lower susceptibility values indicate locations with relatively reduced wildfire likelihood.

The model identifies several concentrated high-susceptibility corridors associated with mountainous terrain, vegetation cover, and historical wildfire activity.

## Summary Statistics

```python
with rasterio.open(
    SUSCEPTIBILITY_PATH
) as src:

    susceptibility = src.read(1)

    nodata = src.nodata

susceptibility = susceptibility[
    susceptibility != nodata
]

susceptibility = susceptibility[
    np.isfinite(
        susceptibility
    )
]
```

```python
summary = pd.DataFrame(
    {
        "Statistic": [
            "Minimum",
            "Maximum",
            "Mean",
            "Median",
            "Standard Deviation"
        ],

        "Value": [
            susceptibility.min(),
            susceptibility.max(),
            susceptibility.mean(),
            np.median(
                susceptibility
            ),
            susceptibility.std()
        ]
    }
)

summary
```

### Summary Statistics Interpretation

The wildfire susceptibility surface ranges from 0.00 to 1.00, representing the full spectrum of predicted wildfire likelihood across Butte County.

The mean susceptibility value of 0.40 indicates that, on average, wildfire potential is relatively low to moderate across the study area. However, the median value of approximately 0.12 is substantially lower than the mean, suggesting that most locations exhibit low susceptibility while a smaller number of highly susceptible areas increase the overall average.

The relatively large standard deviation (0.44) indicates substantial spatial variability in wildfire susceptibility, reflecting strong contrasts between low-risk and high-risk landscapes throughout the county.

## Susceptibility Distribution (Histogram)

```python
plt.figure(figsize=(8,5))

plt.hist(
    susceptibility,
    bins=50
)

plt.xlabel(
    "Susceptibility Index"
)

plt.ylabel(
    "Pixel Count"
)

plt.title(
    "Distribution of Wildfire Susceptibility"
)

plt.tight_layout()

output_path = (
    PROJECT_ROOT /
    "outputs" /
    "figures" /
    "susceptibility_histogram.png"
)

plt.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight"
)

print(
    f"Saved: {output_path}"
)

plt.show()
```

### Susceptibility Distribution Interpretation

The histogram reveals a bimodal distribution of wildfire susceptibility values across Butte County.

A large concentration of pixels occurs near susceptibility values of 0, while another substantial concentration occurs near values of 1. Relatively few pixels fall within intermediate susceptibility ranges.

This pattern indicates that the Random Forest model clearly distinguishes areas with low wildfire potential from areas with high wildfire potential. Rather than producing widespread moderate-risk predictions, the model identifies distinct zones of low and high susceptibility, resulting in strong spatial separation across the landscape.

## Key Findings

- The Random Forest susceptibility model successfully generated a continuous wildfire susceptibility surface ranging from 0 to 1 across Butte County.

- Summary statistics indicate substantial spatial variability in wildfire susceptibility, with most locations exhibiting low susceptibility and a smaller proportion of the landscape exhibiting elevated wildfire potential.

- The bimodal distribution of susceptibility values suggests that the model effectively separates low-risk and high-risk environments, producing clear distinctions between susceptible and non-susceptible areas.

- Areas of elevated susceptibility are concentrated within specific portions of the county rather than being uniformly distributed across the landscape.

- The susceptibility surface provides the foundation for subsequent wildfire exposure and risk assessment analyses.

- For cartographic presentation and decision-support applications, the continuous susceptibility surface will be reclassified into discrete susceptibility categories within ArcGIS Pro. Reclassification improves map readability, facilitates communication with non-technical audiences, and supports interpretation of wildfire hazard patterns by grouping susceptibility values into meaningful classes such as Very Low, Low, Moderate, High, and Very High susceptibility.
