```python
from pathlib import Path

import geopandas as gpd
import matplotlib.pyplot as plt
import rioxarray as rxr
```

```python
DATA_DIR = Path(
    "../data/processed/aligned"
)

BOUNDARY_PATH = (
    "../data/raw/boundaries/boundary.geojson"
)
```

## Load Rasters

```python
dem = rxr.open_rasterio(
    "../data/processed/aligned/dem_aligned.tif"
).squeeze()

lulc = rxr.open_rasterio(
    "../data/processed/aligned/landcover_aligned.tif"
).squeeze()

precip = rxr.open_rasterio(
    "../data/processed/aligned/precipitation_aligned.tif"
).squeeze()
```

## Load Boundary

```python
boundary = gpd.read_file(
    "../data/raw/boundaries/boundary.geojson"
)
```

## DEM Plot

```python
fig, ax = plt.subplots(
    figsize=(8, 6)
)

dem.plot(ax=ax)

boundary.boundary.plot(
    ax=ax,
    color="red",
    linewidth=1
)

ax.set_title(
    "Aligned DEM"
)

plt.show()
```
